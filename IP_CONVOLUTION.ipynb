```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = Image.open("tugas 5 ip.jpeg")

plt.imshow(img)
plt.title("Gambar Asli")
plt.axis("off")
```

```python
gray = img.convert("L")
gray_array = np.array(gray)

plt.imshow(gray_array, cmap="gray")
plt.title("Grayscale")
plt.axis("off")
```

```python
kernel = np.ones((5,5)) / 25
print(kernel)
```

```python
h, w = gray_array.shape
output = np.zeros((h,w))

for i in range(2, h-2):
    for j in range(2, w-2):

        total = 0

        for m in range(-2,3):
            for n in range(-2,3):
                pixel = gray_array[i+m][j+n]
                weight = kernel[m+2][n+2]
                total += pixel * weight

        output[i][j] = total
```

```python
plt.imshow(output, cmap="gray")
plt.title("Hasil Konvolusi Kernel 5x5")
plt.axis("off")
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(gray_array, cmap="gray")
plt.title("Sebelum Konvolusi (Grayscale)")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(output, cmap="gray")
plt.title("Sesudah Konvolusi (Kernel 5x5)")
plt.axis("off")

plt.show()
```
